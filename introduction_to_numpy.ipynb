```python
# Create a 2D array with dimensions (3, 5)
# This means: 3 'rows' and 5 'columns'.
two_d_array = np.full((3, 5), 7) # Example with a specific fill value

print("Resulting 2D array:\n", two_d_array)
print(f"\nShape of the 2D array: {two_d_array.shape}")
```

```python
# Create a 3D array with dimensions (2, 3, 4)
# This means: 2 'blocks' (or matrices), each containing 3 'rows', and each row having 4 'columns'.
three_d_array = np.zeros((2, 3, 4))
print(three_d_array)
print(f"\nShape of the 3D array: {three_d_array.shape}")
```

```python
# Let's create two example 3D arrays
three_d_array_1 = np.ones((2, 3, 4))  # A 3D array of ones
three_d_array_2 = np.zeros((2, 3, 4)) # A 3D array of zeros

print("First 3D array (ones):\n", three_d_array_1)
print("\nSecond 3D array (zeros):\n", three_d_array_2)
```

```python
# Now, stack these two 3D arrays to form a 4D array
# The new dimension will be added at the beginning, so it will have shape (2, 2, 3, 4)
four_d_array_from_3d = np.stack([three_d_array_1, three_d_array_2], axis=0)

print("\nResulting 4D array from stacking 3D arrays:\n", four_d_array_from_3d)
print(f"\nShape of the 4D array: {four_d_array_from_3d.shape}")
```

```python
import numpy as np

def create_2d_array_from_input():
    while True:
        try:
            rows = int(input("Enter the number of rows: "))
            cols = int(input("Enter the number of columns: "))
            break
        except ValueError:
            print("Invalid input. Please enter integers for rows and columns.")

    array_2d = []
    for i in range(rows):
        while True:
            try:
                row_input = input(f"Enter {cols} space-separated integers for row {i+1}: ")
                row = [int(x) for x in row_input.split()]
                if len(row) != cols:
                    raise ValueError(f"Expected {cols} elements, but got {len(row)}.")
                array_2d.append(row)
                break
            except ValueError as e:
                print(f"Invalid input for row {i+1}: {e}. Please try again.")

    return np.array(array_2d)

# Create the 2D array
my_2d_array = create_2d_array_from_input()

print("\nYour 2D array:")
print(my_2d_array)
print(f"Shape of your 2D array: {my_2d_array.shape}")
```
